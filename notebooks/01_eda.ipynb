```python
import pandas as pd 
df = pd.read_csv("../data/creditcard.csv")
df.head()
```

```python
print(df.shape)
```

```python
print(df.columns.tolist())
```

```python
print(df.isnull().sum().sum())
```

```python
print(df["Class"].value_counts())
print()
print(df["Class"].value_counts(normalize=True)*100)
```

```python
import matplotlib.pyplot as plt
df["Class"].value_counts().plot(kind="bar")
plt.title("normal[0] vs fraud[1]")
plt.xlabel("Classes")
plt.ylabel("no of transections")
plt.show()
```

```python
print("NORMAL transactions:")
print(df[df["Class"] == 0]["Amount"].describe())
print()
print("FRAUD transactions:")
print(df[df["Class"] == 1]["Amount"].describe())
```

```python
plt.figure(figsize=(8,4))
plt.hist(df[df["Class"] == 0]["Amount"], bins=50, alpha=0.6, label="Normal", range=(0, 500))
plt.hist(df[df["Class"]==1]["Amount"],bins=50,alpha=0.6,label="Fraud",range=(0,500))
plt.legend()
plt.title("Amount distribution (0 to 500)")
plt.xlabel("Amount")
plt.ylabel("Count")
plt.yscale("log")
plt.show()
```
